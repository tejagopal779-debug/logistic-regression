```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,MinMaxScaler
```

```python
df=pd.read_csv(r"C:\Users\DELL\Downloads\lo\Pass-Fail Data.csv")
```

```python
df.shape
```

```python
df=df.drop(columns="student_id")
```

```python
x=df.drop(columns="pass")
y=df["pass"]
```

```python
mi=MinMaxScaler()
x=mi.fit_transform(x)
```

```python
x_tra, xtes, ytra, ytes=train_test_split(x,y,test_size=0.15,random_state=42)
```

```python
log=LogisticRegression()
u=log.fit(x_tra,ytra)
per=log.predict(xtes)
```

```python
acc=accuracy_score(ytes,per)
acc
```
